## Abstract
This project investigates whether advanced basketball performance metrics, specifically True Shooting Percentage (TS%) and Box Plus-Minus (BPM), can help predict NBA draft selection. As teams start to increasingly rely on data-driven evaluation, understanding which quantitative measures best reflect draft potential is an important question. We focus on TS% as a measure of scoring efficiency and BPM as a measure of overall player impact to determine whether these metrics are associated with a higher likelihood of being drafted.

To address this question, we analyzed a dataset of college basketball players from 2009 to 2021. We conducted exploratory data analysis using visualizations and summary statistics to compare TS%, BPM, and other contextual variables such as player experience and conference affiliation between drafted and non-drafted players. These analyses allowed us to identify patterns and trends in performance metrics and their relationship to draft outcomes.

Our results show that drafted players generally exhibit higher TS% and BPM values, indicating that both efficiency and overall contribution are positively associated with draft selection. However, we also observe significant overlap between drafted and non-drafted players, suggesting that these metrics alone are not sufficient predictors. We conclude that while TS% and BPM are useful indicators of draft potential, NBA draft decisions are influenced by a broader set of factors beyond statistical performance, highlighting the complexity of talent evaluation in professional basketball.

## Background and Prior Work


The NBA Draft is one of the most important mechanisms through which teams acquire young talent, yet predicting which college basketball players will succeed at the professional level remains extremely difficult. Despite teams having access to extensive scouting resources and increasingly detailed performance data, draft outcomes often fail to align with future NBA productivity. This uncertainty has motivated researchers to examine whether measurable college basketball statistics can meaningfully explain if a player is drafted, their downstream success, and whether decision makers systematically rely on the “right” information when making draft selections.

Prior research by Sailofsky (2018) examined NCAA Division I basketball statistics and pre-draft player characteristics to determine which factors predict NBA draft position and future NBA performance.<sup><a href="#ref1">1</a></sup> Using regression models on players drafted between 2006 and 2013, the study found that NBA teams tend to emphasize variables such as scoring totals, size, and college conference affiliation, even though these factors do not strongly predict NBA success. Instead, the paper highlights offensive efficiency, ball control, and rebounding efficiency as more reliable indicators of professional performance. This work suggests that draft decisions are often influenced by cognitive biases and overconfidence, leading teams to prioritize visible or traditional metrics rather than those most closely tied to winning.

Similarily, the article How NBA Teams Use Advanced Stats to Draft Smarter<sup><a href="#ref2">2</a></sup> examines how modern NBA front offices integrate advanced analytics into draft strategy to reduce risk and improve talent evaluation. It explains how teams use predictive modeling built on metrics such as Player Efficiency Rating (PER), Box Plus-Minus (BPM), True Shooting Percentage (TS%), and Win Shares (WS) to project how college and international prospects will translate to the NBA, adjusting for pace, role, and competition level. The article also highlights the growing role of biomechanical analytics, wearable sensors, and motion capture technology in forecasting injury risk, as well as real-time player tracking systems like SportVU and Second Spectrum that capture spatial and decision-making data beyond the box score. Ultimately, the piece emphasizes that while advanced statistics have transformed roster-building by enabling more data-driven projections, the most effective draft strategies blend analytics with traditional scouting to account for intangibles such as character, leadership, and coachability.

In The Life of an Independent NBA Draft Analyst: Behind the Scouting Lens<sup><a href="#ref3">3</a></sup>, Rafael Barlowe provides a firsthand account of the realities of independent NBA draft scouting, highlighting the balance between in-person evaluation, film study, and navigating the politics of draft intel. He describes the unpredictability of prospect evaluation and the limitations of relying solely on game film, emphasizing that live scouting reveals nuances such as demeanor, communication, and competitiveness that statistics cannot capture. Barlowe outlines a rigorous evaluation process that combines full-game film breakdowns, advanced analytics, role projection, and possession-level review to assess how prospects might function within NBA ecosystems. He also reflects on the importance of trusting independent judgment over consensus, learning from scouting misses, and accounting for off-court factors such as maturity, cultural adjustment, and support systems. This source helped us understand the limitations of our desired project, since it is nearly impossible to model qualitative factors such as effort and maturity, which will lead to a characteristic gap in our eventual analysis.

Building on this prior work, our project focuses specifically on examining the statistical relationships between college basketball performance metrics and NBA draft outcomes for players eligible for the NBA Draft. Our analysis aims to identify patterns and associations in the data, contributing to existing research on draft decision-making and highlighting which types of college-level statistics appear most closely linked to draft selection.

## Hypothesis


We hypothesize that higher True Shooting Percentage (TS%, defined as a player’s scoring efficiency accounting for field goals, three-pointers, and free throws) and BPM (Box Plus Minus) values are positively associated with NBA draft selection. Specifically, we expect that players with higher efficiency and overall impact metrics will have a greater probability of being drafted (Drafted = 1).

We believe TS% and BPM are particularly important because they capture overall scoring efficiency and a player’s total on-court impact, rather than raw scoring volume alone. Unlike traditional statistics that reflect usage or opportunity, efficiency-based metrics better account for shot quality, decision-making, and two-way contribution, which may be more indicative of NBA-ready performance. Additionally, we expect player characteristics such as position, and participation in higher-profile conferences to be positively associated with draft selection.

### Analysis: Advanced Metrics Two-Sample t-tests

In this section, we use two-sample t-tests to evaluate whether there are statistically significant differences in advanced performance metrics—specifically True Shooting Percentage (TS%) and Box Plus-Minus (BPM)—between players who were drafted and those who were not. By comparing the mean values of these metrics across the two groups, we aim to determine whether observed differences in efficiency and overall player impact are likely due to random variation or reflect meaningful distinctions related to draft selection. The t-test allows us to formally assess the strength of these relationships using p-values, providing a more rigorous statistical foundation beyond the visual trends observed in our exploratory data analysis.

```python
# null hypothesis = average TS% for drafted players and undrafted players are the same
# alternate hypothesis = average TS% for drafted players is higher than for undrafted players

drafted_df = cleaned_df[cleaned_df["drafted"] == 1]
undrafted_df = cleaned_df[cleaned_df["drafted"] == 0]

print(f"Average TS% for drafted players: {drafted_df["TS_per"].mean()}")
print(f"Average TS% for undrafted players: {undrafted_df["TS_per"].mean()}")

from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(
    drafted_df["TS_per"],
    undrafted_df["TS_per"],
    equal_var=False
)

print(f"T-test statistic: {t_stat}")
print(f"P-value: {p_value}")
```

```python
# null hypothesis = average BPM for drafted players and undrafted players are the same
# alternate hypothesis = average BPM for drafted players is higher than for undrafted players

print(f"Average BPM for drafted players: {drafted_df["bpm"].mean()}")
print(f"Average BPM for undrafted players: {undrafted_df["bpm"].mean()}")


t_stat, p_value = ttest_ind(
    drafted_df["bpm"],
    undrafted_df["bpm"],
    equal_var=False
)

print(f"T-test statistic: {t_stat}")
print(f"P-value: {p_value}")
```

### Analysis 2: Logistic Regression model with gridsearchcv

In this section, we implement a logistic regression model to predict NBA draft selection based on advanced performance metrics, including True Shooting Percentage (TS%) and Box Plus-Minus (BPM), along with other relevant features such as player experience and conference. Logistic regression is appropriate for this task because the outcome variable—whether a player is drafted—is binary. To improve model performance and select optimal hyperparameters, we use GridSearchCV to systematically evaluate different configurations and identify the best-performing model based on cross-validation. This approach allows us to not only assess the predictive power of these features but also understand how well advanced metrics can generalize in distinguishing between drafted and non-drafted players.

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report

numeric_features = ["TS_per", "bpm"]
categorical_features = ["conf", "yr_numerical"]

X = cleaned_df[numeric_features + categorical_features]
y = cleaned_df["drafted"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

numeric_transformer = StandardScaler()
categorical_transformer = OneHotEncoder(handle_unknown="ignore")

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])

param_grid = {
    "model__C": [0.01, 0.1, 1, 10],
    "model__penalty": ["l1", "l2"],
    "model__solver": ["lbfgs", "liblinear"]
}

grid_search = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)
print("Best CV score:", grid_search.best_score_)

best_model = grid_search.best_estimator_

y_pred = best_model.predict(X_test)

print(classification_report(y_test, y_pred))


model = best_model.named_steps["model"]
feature_names = best_model.named_steps["preprocessor"].get_feature_names_out()
coef_df = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": model.coef_[0]
})

coef_df = coef_df.sort_values(by="Coefficient", ascending=False).reset_index(drop=True)

print(coef_df.head(10))

coef_df.sort_values("Coefficient").plot(
    x="Feature",
    y="Coefficient",
    kind="barh",
    figsize=(10,8)
)

plt.title("Feature Importance from Logistic Regression")
plt.xlabel("Coefficient")
plt.show()
```

Beyond individual performance metrics, our analysis revealed that conference affiliation plays a notable role in draft outcomes. The ACC, Pac-10, and SEC had draft rates of approximately 13–14%, nearly 5× the overall dataset draft rate of 2.7%. In fact, only 25% of all NCAA conferences exceeded the overall draft rate, and the median conference draft rate was just 0.44% — well below the average. This suggests that players from high-profile conferences receive substantially more exposure to NBA scouts, and that conference affiliation may act as an independent signal of draft potential beyond raw statistical performance alone.

Building on these findings, we implemented a logistic regression model with GridSearchCV to evaluate the predictive power of these features. BPM emerged as the strongest single predictor, carrying the largest positive coefficient in the model, while TS% also contributed positively. Senior year (year = 4) had a strongly negative coefficient, consistent with our EDA finding that freshmen are drafted at roughly 3× the rate of seniors. Although the model achieved high precision (~0.98) and recall (~1.00) for the non-drafted class, recall for the drafted class was noticeably lower, indicating that while advanced metrics provide useful signals, they do not fully explain draft selection on their own.

There are several important limitations to our study. Most critically, our dataset is heavily imbalanced — only 2.7% of the 22,393 player-season observations represent drafted players, meaning there are roughly 36 non-drafted players for every 1 drafted player. This extreme imbalance makes it inherently difficult for the model to identify the drafted class, which explains the lower recall on class 1 despite overall high accuracy. Additionally, the dataset lacks key variables that scouts weigh heavily, such as height, wingspan, injury history, athleticism, and team context. Our analysis is also based on historical data from 2009–2021 and does not account for shifts in scouting strategy or how the modern NBA's style of play has evolved over time. The presence of significant overlap between drafted and non-drafted players in our metrics further confirms that statistical performance alone cannot fully differentiate between the two groups.

For future work, we could expand this analysis by incorporating physical attributes such as height and wingspan, team-level factors, and longitudinal data to better capture player development across multiple seasons. Exploring more complex models such as random forests or neural networks may also improve recall on the drafted class. Additionally, predicting draft round or draft position — rather than just a binary drafted/not-drafted outcome — could provide deeper insight into how performance metrics influence where players are selected. Overall, this project demonstrates that while TS%, BPM, and conference affiliation are valuable and statistically significant indicators of draft potential, predicting NBA draft outcomes remains a complex problem that requires a broader set of variables and modeling approaches.
